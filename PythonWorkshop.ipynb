```python
#4
num1 = 25
num2 = 10

print(f"Numbers chosen: num1 = {num1}, num2 = {num2}\n")

print("--- Arithmetic Operations ---")
print(f"Addition ({num1} + {num2}): {num1 + num2}")
print(f"Subtraction ({num1} - {num2}): {num1 - num2}")
print(f"Multiplication ({num1} * {num2}): {num1 * num2}")
print(f"Division ({num1} / {num2}): {num1 / num2}")
print(f"Floor Division ({num1} // {num2}): {num1 // num2}")
print(f"Modulus ({num1} % {num2}): {num1 % num2}")
print(f"Exponentiation ({num1} ** 2): {num1 ** 2}\n")

print("--- Comparison Operations ---")
print(f"Equal to ({num1} == {num2}): {num1 == num2}")
print(f"Not Equal to ({num1} != {num2}): {num1 != num2}")
print(f"Greater Than ({num1} > {num2}): {num1 > num2}")
print(f"Less Than ({num1} < {num2}): {num1 < num2}")
print(f"Greater Than or Equal To ({num1} >= {num2}): {num1 >= num2}")
print(f"Less Than or Equal To ({num1} <= {num2}): {num1 <= num2}\n")

print("--- Logical Operations ---")
is_positive_num1 = (num1 > 0)
is_positive_num2 = (num2 > 0)
is_even_num1 = (num1 % 2 == 0)

print(f"num1 > 0 (True) AND num2 > 0 (True): {is_positive_num1 and is_positive_num2}")
print(f"num1 > 0 (True) OR num1 is even (False): {is_positive_num1 or is_even_num1}")
print(f"NOT (num1 is even (False)): {not is_even_num1}")
```

```python
user_string = input("Enter a string: ")
print(f"You entered: '{user_string}'\n")

second_string = " Python is fun!"
concatenated_string = user_string + second_string
print(f"Concatenation: '{concatenated_string}'\n")

if len(user_string) > 0:
    print(f"First character (index 0): '{user_string[0]}'")
    print(f"Last character (index -1): '{user_string[-1]}'\n")
else:
    print("String is empty, cannot perform indexing.\n")

if len(user_string) > 3:
    sliced_string = user_string[1:4]
    print(f"Slicing (characters from index 1 to 3): '{sliced_string}'\n")
else:
    print("String is too short for slicing example.\n")

uppercase_string = user_string.upper()
print(f"Uppercase: '{uppercase_string}'")

lowercase_string = user_string.lower()
print(f"Lowercase: '{lowercase_string}'")

split_list = user_string.split()
print(f"Split by whitespace: {split_list}")

replaced_string = user_string.replace('e', 'X')
print(f"Replace 'e' with 'X': '{replaced_string}'")

string_length = len(user_string)
print(f"Length of the string: {string_length}")
```

```python
#6 --- Python Program: Determine Student Status (Pass/Fail/Re-test) ---

try:
    marks = float(input("Enter the student's marks (out of 100): "))
except ValueError:
    print("Invalid input. Please enter a numeric value for marks.")
    marks = -1

if marks >= 0 and marks <= 100:
    if marks >= 50:
        print(f"With {marks} marks, the student has: Pass")
    elif marks >= 40:
        print(f"With {marks} marks, the student is: Eligible for Re-test")
    else:
        print(f"With {marks} marks, the student has: Fail")
elif marks != -1:
    print("Marks should be between 0 and 100.")
```

```python
#7 Python Program: Check Even or Odd Number

try:
    number = int(input("Enter an integer: "))
except ValueError:
    print("Invalid input. Please enter an integer.")
else:
    if number % 2 == 0:
        print(f"The number {number} is Even.")
    else:
        print(f"The number {number} is Odd.")
```

```python
#9
num_rows = int(input("Enter the number of rows for the star pattern: "))

print("\n--- Star Pattern with break and continue ---")

for i in range(1, num_rows + 1):
    if i > 5:
        print("Max rows for demonstration reached. Breaking outer loop.")
        break
    for j in range(1, i + 1):
        if i == 3 and j == 2:
            print(" (skipped)", end="")
            continue
        print("*", end="")
    print()
```

```python
num_rows_simple = int(input("Enter the number of rows for the simple star pattern: "))

print("\n--- Simple Star Pattern with pass ---")

for i in range(1, num_rows_simple + 1):
    for j in range(1, i + 1):
        if i == 2 and j == 1:
            pass
        print("*", end="")
    print()
```
