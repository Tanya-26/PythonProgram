```python
# --- Python Program: Variables, Naming Conventions, and Data Types ---

# 1. Variable Naming Conventions
#    - Use descriptive names (e.g., 'user_age' instead of 'ua').
#    - Use snake_case for variable names (all lowercase, words separated by underscores).
#    - Avoid single-letter variable names unless it's a counter or very temporary.
#    - Variable names cannot start with a number.
#    - Variable names can only contain alpha-numeric characters and underscores (A-z, 0-9, and _).
#    - Python keywords cannot be used as variable names (e.g., 'if', 'for', 'class').

# 2. Demonstrating Basic Data Types

# Integer (int): Whole numbers, positive or negative, without decimals.
user_age = 30
product_quantity = 150
print(f"Integer Variable 'user_age': {user_age}, Type: {type(user_age)}")
print(f"Integer Variable 'product_quantity': {product_quantity}, Type: {type(product_quantity)}\n")

# Float (float): Numbers with a decimal point.
item_price = 29.99
temperature_celsius = 25.5
print(f"Float Variable 'item_price': {item_price}, Type: {type(item_price)}")
print(f"Float Variable 'temperature_celsius': {temperature_celsius}, Type: {type(temperature_celsius)}\n")

# String (str): A sequence of characters, enclosed in single or double quotes.
user_name = "Alice Smith"
welcome_message = 'Hello, Python world!'
print(f"String Variable 'user_name': {user_name}, Type: {type(user_name)}")
print(f"String Variable 'welcome_message': {welcome_message}, Type: {type(welcome_message)}\n")

# Boolean (bool): Represents truth values, either True or False.
is_admin = True
has_permission = False
print(f"Boolean Variable 'is_admin': {is_admin}, Type: {type(is_admin)}")
print(f"Boolean Variable 'has_permission': {has_permission}, Type: {type(has_permission)}\n")

# You can also reassign variables, changing their value and potentially their type.
# This is generally discouraged for clarity if not intended, but possible.
example_variable = 10
print(f"Initial 'example_variable': {example_variable}, Type: {type(example_variable)}")
example_variable = "Now a string"
print(f"Reassigned 'example_variable': {example_variable}, Type: {type(example_variable)}")
```

```python
# --- Python Program: Type Casting ---

# 1. Integer to Float Conversion
int_num = 10
float_from_int = float(int_num)
print(f"Original integer: {int_num} (Type: {type(int_num)})")
print(f"Converted to float: {float_from_int} (Type: {type(float_from_int)})\n")

# 2. Float to Integer Conversion
float_num = 15.75
int_from_float = int(float_num)
print(f"Original float: {float_num} (Type: {type(float_num)})")
print(f"Converted to integer: {int_from_float} (Type: {type(int_from_float)}) (Note: decimal part is truncated)\n")

# 3. Integer to String Conversion
num_to_str = 123
string_from_int = str(num_to_str)
print(f"Original integer: {num_to_str} (Type: {type(num_to_str)})")
print(f"Converted to string: '{string_from_int}' (Type: {type(string_from_int)})\n")

# 4. String to Integer Conversion (if valid numeric string)
str_num = "456"
int_from_str = int(str_num)
print(f"Original string: '{str_num}' (Type: {type(str_num)})")
print(f"Converted to integer: {int_from_str} (Type: {type(int_from_str)})\n")

# 5. String to Float Conversion (if valid numeric string)
str_float = "78.90"
float_from_str = float(str_float)
print(f"Original string: '{str_float}' (Type: {type(str_float)})")
print(f"Converted to float: {float_from_str} (Type: {type(float_from_str)})\n")

# --- Important Note on Type Casting ---
# Be cautious when converting between types, especially when converting strings
# to numbers. If the string does not contain a valid number, a ValueError will occur.
```

```python
# --- Python Program: Arithmetic, Comparison, and Logical Operations ---

# Define two numbers for demonstration
num1 = 25
num2 = 10

print(f"Numbers chosen: num1 = {num1}, num2 = {num2}\n")

# 1. Arithmetic Operations
print("--- Arithmetic Operations ---")
print(f"Addition ({num1} + {num2}): {num1 + num2}")
print(f"Subtraction ({num1} - {num2}): {num1 - num2}")
print(f"Multiplication ({num1} * {num2}): {num1 * num2}")
print(f"Division ({num1} / {num2}): {num1 / num2}")
print(f"Floor Division ({num1} // {num2}): {num1 // num2}")
print(f"Modulus ({num1} % {num2}): {num1 % num2}")
print(f"Exponentiation ({num1} ** 2): {num1 ** 2}\n")

# 2. Comparison Operations (Output is always a Boolean: True or False)
print("--- Comparison Operations ---")
print(f"Equal to ({num1} == {num2}): {num1 == num2}")
print(f"Not Equal to ({num1} != {num2}): {num1 != num2}")
print(f"Greater Than ({num1} > {num2}): {num1 > num2}")
print(f"Less Than ({num1} < {num2}): {num1 < num2}")
print(f"Greater Than or Equal To ({num1} >= {num2}): {num1 >= num2}")
print(f"Less Than or Equal To ({num1} <= {num2}): {num1 <= num2}\n")

# 3. Logical Operations (Typically used with Boolean values)
print("--- Logical Operations ---")
is_positive_num1 = (num1 > 0)
is_positive_num2 = (num2 > 0)
is_even_num1 = (num1 % 2 == 0)

print(f"num1 > 0 (True) AND num2 > 0 (True): {is_positive_num1 and is_positive_num2}")
print(f"num1 > 0 (True) OR num1 is even (False): {is_positive_num1 or is_even_num1}")
print(f"NOT (num1 is even (False)): {not is_even_num1}")
```

```python
# --- Python Program: String Operations ---

# 1. Take Input from the User
user_string = input("Enter a string: ")
print(f"You entered: '{user_string}'\n")

# 2. String Concatenation
second_string = " Python is fun!"
concatenated_string = user_string + second_string
print(f"Concatenation: '{concatenated_string}'\n")

# 3. String Indexing
if len(user_string) > 0:
    print(f"First character (index 0): '{user_string[0]}'")
    print(f"Last character (index -1): '{user_string[-1]}'\n")
else:
    print("String is empty, cannot perform indexing.\n")

# 4. String Slicing
if len(user_string) > 3:
    sliced_string = user_string[1:4] # Characters from index 1 up to (but not including) 4
    print(f"Slicing (characters from index 1 to 3): '{sliced_string}'\n")
else:
    print("String is too short for slicing example.\n")

# 5. Common String Methods

# upper(): Converts string to uppercase
uppercase_string = user_string.upper()
print(f"Uppercase: '{uppercase_string}'")

# lower(): Converts string to lowercase
lowercase_string = user_string.lower()
print(f"Lowercase: '{lowercase_string}'")

# split(): Splits the string by a delimiter (default is whitespace)
split_list = user_string.split()
print(f"Split by whitespace: {split_list}")

# replace(): Replaces occurrences of a substring
# Let's replace 'e' with 'X' as an example
replaced_string = user_string.replace('e', 'X')
print(f"Replace 'e' with 'X': '{replaced_string}'")

# len(): Returns the length of the string
string_length = len(user_string)
print(f"Length of the string: {string_length}")
```
